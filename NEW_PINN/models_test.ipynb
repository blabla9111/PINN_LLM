```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import os
from pathlib import Path
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import json
from datetime import datetime

# ============================================================
# НАСТРОЙКА
# ============================================================

# Определяем девайс
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Фиксируем seed для воспроизводимости
torch.manual_seed(123)
if device.type == 'cuda':
    torch.cuda.manual_seed(123)
np.random.seed(123)
torch.backends.cudnn.deterministic = True

# Создаем папку для результатов
results_dir = Path("synthetic_datasets_results")
results_dir.mkdir(exist_ok=True)

# Создаем подпапки для каждой модели
model1_dir = results_dir / "model1_constant"
model2_dir = results_dir / "model2_temporal"
model1_dir.mkdir(exist_ok=True)
model2_dir.mkdir(exist_ok=True)

# ============================================================
# СПИСОК ДАТАСЕТОВ ДЛЯ ТЕСТИРОВАНИЯ
# ============================================================

datasets = [
    # Основные датасеты
    {"name": "01_baseline_constant", "desc": "Базовый (константы)", "has_noise": False},
    {"name": "02_lockdown_beta_jump", "desc": "Локдаун (скачок β)", "has_noise": False},
    {"name": "03_seasonal_beta_sin", "desc": "Сезонность (плавная β)", "has_noise": False},
    {"name": "04_decaying_beta_trend", "desc": "Затухание (β 0.15→0.1)", "has_noise": False},
    {"name": "05_complex_beta_gamma_dynamics", "desc": "Сложный (β+γ)", "has_noise": False},
    
    # Шумные версии датасета 2
    {"name": "02_lockdown_beta_jump_noise_3pct_seed42", "desc": "Локдаун + шум 3%", "has_noise": True},
    {"name": "02_lockdown_beta_jump_noise_5pct_seed123", "desc": "Локдаун + шум 5%", "has_noise": True},
    {"name": "02_lockdown_beta_jump_noise_10pct_seed456", "desc": "Локдаун + шум 10%", "has_noise": True},
    
    # Шумные версии датасета 5
    {"name": "05_complex_beta_gamma_dynamics_noise_3pct_seed42", "desc": "Сложный + шум 3%", "has_noise": True},
    {"name": "05_complex_beta_gamma_dynamics_noise_5pct_seed123", "desc": "Сложный + шум 5%", "has_noise": True},
    {"name": "05_complex_beta_gamma_dynamics_noise_10pct_seed456", "desc": "Сложный + шум 10%", "has_noise": True},
]

# Истинные параметры для каждого датасета (для расчета ошибок)
true_params = {
    "01_baseline_constant": {"beta": 0.1, "gamma": 0.06, "mu": 0.003, "type": "constant"},
    "02_lockdown_beta_jump": {"beta": "time_varying", "gamma": 0.06, "mu": 0.003, "type": "beta_time"},
    "03_seasonal_beta_sin": {"beta": "time_varying", "gamma": 0.06, "mu": 0.003, "type": "beta_time"},
    "04_decaying_beta_trend": {"beta": "time_varying", "gamma": 0.06, "mu": 0.003, "type": "beta_time"},
    "05_complex_beta_gamma_dynamics": {"beta": "time_varying", "gamma": "time_varying", "mu": 0.003, "type": "both_time"},
}

# Для шумных версий параметры те же, что у исходных
for noise_level in [3, 5, 10]:
    for seed in [42, 123, 456]:
        true_params[f"02_lockdown_beta_jump_noise_{noise_level}pct_seed{seed}"] = true_params["02_lockdown_beta_jump"]
        true_params[f"05_complex_beta_gamma_dynamics_noise_{noise_level}pct_seed{seed}"] = true_params["05_complex_beta_gamma_dynamics"]

# ============================================================
# ФУНКЦИИ ДЛЯ РАСЧЕТА МЕТРИК
# ============================================================

def calculate_metrics(y_true, y_pred):
    """Рассчитывает MSE, RMSE, MAE для временных рядов"""
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    return {"MSE": mse, "RMSE": rmse, "MAE": mae}

def calculate_parameter_error_model1(pred_params, true_params_dict):
    """Относительная ошибка для константных параметров (модель 1)"""
    errors = {}
    for param in ['beta', 'gamma', 'mu']:
        if isinstance(true_params_dict[param], (int, float)):
            true_val = true_params_dict[param]
            pred_val = pred_params[param]
            rel_error = abs(pred_val - true_val) / true_val
            errors[param] = {
                "true": true_val,
                "pred": pred_val,
                "abs_error": abs(pred_val - true_val),
                "rel_error": rel_error,
                "rel_error_percent": rel_error * 100
            }
    return errors

def calculate_parameter_error_model2(df_true, df_pred, param_name):
    """Ошибки для временных параметров (модель 2)"""
    true_vals = df_true[param_name].values
    pred_vals = df_pred[param_name].values
    
    # Обрезаем до одинаковой длины
    min_len = min(len(true_vals), len(pred_vals))
    true_vals = true_vals[:min_len]
    pred_vals = pred_vals[:min_len]
    
    mse = mean_squared_error(true_vals, pred_vals)
    rmse = np.sqrt(mse)
    r2 = r2_score(true_vals, pred_vals)
    
    return {
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2,
        "mean_true": np.mean(true_vals),
        "mean_pred": np.mean(pred_vals)
    }

# ============================================================
# ФУНКЦИИ ДЛЯ ВИЗУАЛИЗАЦИИ
# ============================================================

def plot_model_results(df_true, df_pred_model1, df_pred_model2, dataset_name, save_dir):
    """Создает визуализацию сравнения моделей"""
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    fig.suptitle(f'Сравнение моделей: {dataset_name}', fontsize=16, fontweight='bold')
    
    t = df_true['day'].values
    
    # 1. Инфицированные I(t)
    ax = axes[0, 0]
    ax.plot(t, df_true['I'], 'k-', linewidth=2.5, label='Истинные', alpha=0.8)
    ax.plot(t[:len(df_pred_model1['I'])], df_pred_model1['I'], 'r--', linewidth=2, label='Модель 1 (конст)')
    ax.plot(t[:len(df_pred_model2['I'])], df_pred_model2['I'], 'b--', linewidth=2, label='Модель 2 (врем)')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Инфицированные')
    ax.set_title('Динамика I(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 2. Умершие D(t)
    ax = axes[0, 1]
    ax.plot(t, df_true['D'], 'k-', linewidth=2.5, label='Истинные', alpha=0.8)
    ax.plot(t[:len(df_pred_model1['D'])], df_pred_model1['D'], 'r--', linewidth=2, label='Модель 1')
    ax.plot(t[:len(df_pred_model2['D'])], df_pred_model2['D'], 'b--', linewidth=2, label='Модель 2')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Умершие')
    ax.set_title('Динамика D(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 3. R₀ (если есть в данных)
    ax = axes[0, 2]
    if 'R0' in df_true.columns:
        ax.plot(t, df_true['R0'], 'k-', linewidth=2.5, label='Истинный R₀', alpha=0.8)
    if 'R0' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['R0'])], df_pred_model2['R0'], 'b--', linewidth=2, label='Модель 2 R₀')
    ax.axhline(y=1.0, color='gray', linestyle='--', alpha=0.5)
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('R₀')
    ax.set_title('Репродуктивное число')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 4. Параметр β(t)
    ax = axes[1, 0]
    if 'beta' in df_true.columns:
        ax.plot(t, df_true['beta'], 'k-', linewidth=2.5, label='Истинный β(t)', alpha=0.8)
    if 'beta' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['beta'])], df_pred_model2['beta'], 'b--', linewidth=2, label='Модель 2 β(t)')
    # Для модели 1 - константа
    if 'beta_const' in df_pred_model1.columns:
        beta_const = df_pred_model1['beta_const'].iloc[0]
        ax.axhline(y=beta_const, color='r', linestyle='--', linewidth=2, label=f'Модель 1 β={beta_const:.4f}')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('β')
    ax.set_title('Параметр β(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 5. Параметры γ(t) и μ(t)
    ax = axes[1, 1]
    if 'gamma' in df_true.columns:
        ax.plot(t, df_true['gamma'], 'g-', linewidth=2, label='Истинный γ(t)', alpha=0.6)
    if 'mu' in df_true.columns:
        ax.plot(t, df_true['mu'], 'brown', linewidth=2, label='Истинный μ(t)', alpha=0.6)
    if 'gamma' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['gamma'])], df_pred_model2['gamma'], 'g--', linewidth=2, label='Модель 2 γ(t)')
    if 'mu' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['mu'])], df_pred_model2['mu'], 'orange', linewidth=2, label='Модель 2 μ(t)')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Параметры')
    ax.set_title('Параметры γ(t) и μ(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 6. Ошибки предсказания I(t)
    ax = axes[1, 2]
    min_len = min(len(t), len(df_pred_model1['I']), len(df_pred_model2['I']))
    error1 = np.abs(df_true['I'].values[:min_len] - df_pred_model1['I'][:min_len])
    error2 = np.abs(df_true['I'].values[:min_len] - df_pred_model2['I'][:min_len])
    
    ax.plot(t[:min_len], error1, 'r-', linewidth=1.5, label='Ошибка Модель 1')
    ax.plot(t[:min_len], error2, 'b-', linewidth=1.5, label='Ошибка Модель 2')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('|Ошибка| I(t)')
    ax.set_title('Абсолютная ошибка для I(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.savefig(save_dir / f"{dataset_name}_comparison.png", dpi=150, bbox_inches='tight')
    plt.close()

# ============================================================
# ОСНОВНОЙ ЦИКЛ ТЕСТИРОВАНИЯ
# ============================================================

def run_testing():
    """Запускает тестирование на всех датасетах"""
    
    # Словари для накопления результатов
    results_model1 = []
    results_model2 = []
    
    print("="*80)
    print("НАЧАЛО ТЕСТИРОВАНИЯ МОДЕЛЕЙ")
    print("="*80)
    
    for ds in datasets:
        dataset_name = ds["name"]
        dataset_desc = ds["desc"]
        
        print(f"\n{'='*60}")
        print(f"ТЕСТИРОВАНИЕ: {dataset_name}")
        print(f"Описание: {dataset_desc}")
        print('='*60)
        
        try:
            # ==================================================
            # ЗАГРУЗКА ДАННЫХ
            # ==================================================
            file_path = f"./synthetic_datasets/{dataset_name}.csv"
            if not os.path.exists(file_path):
                print(f"❌ Файл не найден: {file_path}")
                continue
                
            df_true = pd.read_csv(file_path)
            
            # Получаем данные
            S = df_true['S'].values
            I = df_true['I'].values
            D = df_true['D'].values
            R = df_true['R'].values
            timesteps = np.arange(len(S)).astype(float)
            
            # Параметры
            # train_size = min(101, len(timesteps) // 2)  # Обучаем на половине данных
            train_size = 50
            population = 1000
            
            print(f"Всего точек: {len(timesteps)}, Обучающих: {train_size}, Тестовых: {len(timesteps)-train_size}")
            
            # ==================================================
            # МОДЕЛЬ 1: EINN_PINN (константные параметры)
            # ==================================================
            print("\n--- МОДЕЛЬ 1 (константные параметры) ---")
            
            from NEW_PINN.PINN_const import EINN_PINN  # предполагаемый импорт
            
            model1 = EINN_PINN(
                t=timesteps,
                S_data=S,
                I_data=I,
                R_data=R,
                D_data=D,
                population=population,
                train_size=train_size,
                device=device
            )
            
            # Обучаем
            model1.train_model(n_epoch=20000, lambda_data=0.01, lambda_ode=1.0)
            
            # Получаем параметры
            final_params = model1.params.get_params_dict()
            
            # Предсказание
            S_pred1, I_pred1, R_pred1, D_pred1 = model1.predict(timesteps)
            
            # Конвертируем в numpy
            S_pred1 = S_pred1.numpy()
            I_pred1 = I_pred1.numpy()
            R_pred1 = R_pred1.numpy()
            D_pred1 = D_pred1.numpy()
            
            # Создаем DataFrame с предсказаниями модели 1
            df_pred1 = pd.DataFrame({
                'day': timesteps[:len(S_pred1)],
                'S': S_pred1,
                'I': I_pred1,
                'R': R_pred1,
                'D': D_pred1,
                'beta_const': final_params['beta'],
                'gamma_const': final_params['gamma'],
                'mu_const': final_params['mu']
            })
            df_pred1['R0_const'] = df_pred1['beta_const'] / (df_pred1['gamma_const'] + df_pred1['mu_const'])
            
            # ==================================================
            # МОДЕЛЬ 2: Nature_PINN (временные параметры)
            # ==================================================
            print("\n--- МОДЕЛЬ 2 (временные параметры) ---")
            
            from NEW_PINN.PINN_time import Nature_PINN  # предполагаемый импорт
            
            model2 = Nature_PINN(
                t=timesteps,
                S_data=S,
                I_data=I,
                D_data=D,
                R_data=R,
                device=device,
                train_size=train_size,
                population_n=population
            )
            
            # Обучаем
            model2.train(n_epoch=20000)
            
            # Получаем предсказания
            predictions = model2.predict()
            
            # Создаем DataFrame с предсказаниями модели 2
            df_pred2 = pd.DataFrame({
                'day': predictions['t'][:len(predictions['S'])],
                'S': predictions['S'],
                'I': predictions['I'],
                'R': predictions['R'],
                'D': predictions['D'],
                'beta': predictions['beta'],
                'gamma': predictions['gamma'],
                'mu': predictions['mu']
            })
            df_pred2['R0'] = df_pred2['beta'] / (df_pred2['gamma'] + df_pred2['mu'])
            
            # ==================================================
            # РАСЧЕТ МЕТРИК
            # ==================================================
            
            # Обрезаем до одинаковой длины
            min_len = min(len(df_true), len(df_pred1), len(df_pred2))
            df_true_trim = df_true.iloc[:min_len]
            df_pred1_trim = df_pred1.iloc[:min_len]
            df_pred2_trim = df_pred2.iloc[:min_len]
            
            # Метрики для SIRD
            metrics_model1 = {}
            metrics_model2 = {}
            
            for compartment in ['S', 'I', 'R', 'D']:
                true_vals = df_true_trim[compartment].values
                pred1_vals = df_pred1_trim[compartment].values
                pred2_vals = df_pred2_trim[compartment].values
                
                metrics_model1[compartment] = calculate_metrics(true_vals, pred1_vals)
                metrics_model2[compartment] = calculate_metrics(true_vals, pred2_vals)
            
            # Ошибки параметров для модели 1
            param_errors_model1 = {}
            if dataset_name in true_params:
                if true_params[dataset_name]["type"] == "constant":
                    param_errors_model1 = calculate_parameter_error_model1(final_params, true_params[dataset_name])
            
            # Ошибки параметров для модели 2
            param_errors_model2 = {}
            if dataset_name in true_params:
                if 'beta' in df_true.columns and true_params[dataset_name]["beta"] == "time_varying":
                    param_errors_model2['beta'] = calculate_parameter_error_model2(df_true, df_pred2, 'beta')
                if 'gamma' in df_true.columns and true_params[dataset_name]["gamma"] == "time_varying":
                    param_errors_model2['gamma'] = calculate_parameter_error_model2(df_true, df_pred2, 'gamma')
                if 'mu' in df_true.columns and isinstance(true_params[dataset_name]["mu"], (int, float)):
                    param_errors_model2['mu'] = calculate_parameter_error_model2(df_true, df_pred2, 'mu')
            
            # ==================================================
            # СОХРАНЕНИЕ РЕЗУЛЬТАТОВ
            # ==================================================
            
            # Сохраняем предсказания
            df_pred1.to_csv(model1_dir / f"{dataset_name}_predictions.csv", index=False)
            df_pred2.to_csv(model2_dir / f"{dataset_name}_predictions.csv", index=False)
            
            # Сохраняем метрики
            result_entry_model1 = {
                'dataset': dataset_name,
                'description': dataset_desc,
                **{f"{comp}_MSE": metrics_model1[comp]['MSE'] for comp in ['S', 'I', 'R', 'D']},
                **{f"{comp}_RMSE": metrics_model1[comp]['RMSE'] for comp in ['S', 'I', 'R', 'D']},
                **{f"{comp}_MAE": metrics_model1[comp]['MAE'] for comp in ['S', 'I', 'R', 'D']},
                'beta_pred': final_params['beta'],
                'gamma_pred': final_params['gamma'],
                'mu_pred': final_params['mu'],
            }
            
            # Добавляем ошибки параметров если есть
            if param_errors_model1:
                for param, errors in param_errors_model1.items():
                    result_entry_model1[f"{param}_true"] = errors['true']
                    result_entry_model1[f"{param}_rel_error"] = errors['rel_error']
                    result_entry_model1[f"{param}_rel_error_percent"] = errors['rel_error_percent']
            
            results_model1.append(result_entry_model1)
            
            result_entry_model2 = {
                'dataset': dataset_name,
                'description': dataset_desc,
                **{f"{comp}_MSE": metrics_model2[comp]['MSE'] for comp in ['S', 'I', 'R', 'D']},
                **{f"{comp}_RMSE": metrics_model2[comp]['RMSE'] for comp in ['S', 'I', 'R', 'D']},
                **{f"{comp}_MAE": metrics_model2[comp]['MAE'] for comp in ['S', 'I', 'R', 'D']},
            }
            
            # Добавляем ошибки параметров
            for param, errors in param_errors_model2.items():
                result_entry_model2[f"{param}_RMSE"] = errors['RMSE']
                result_entry_model2[f"{param}_R2"] = errors['R2']
            
            results_model2.append(result_entry_model2)
            
            # ==================================================
            # ВИЗУАЛИЗАЦИЯ
            # ==================================================
            
            plot_model_results(df_true, df_pred1, df_pred2, dataset_name, results_dir)
            
            print(f"\n✅ Завершено: {dataset_name}")
            print(f"   Модель 1 - I_RMSE: {metrics_model1['I']['RMSE']:.4f}")
            print(f"   Модель 2 - I_RMSE: {metrics_model2['I']['RMSE']:.4f}")
            
        except Exception as e:
            print(f"❌ Ошибка при обработке {dataset_name}: {str(e)}")
            import traceback
            traceback.print_exc()
            continue
    
    # ==================================================
    # СОХРАНЕНИЕ СВОДНЫХ РЕЗУЛЬТАТОВ
    # ==================================================
    
    # Сохраняем результаты в CSV
    df_results_model1 = pd.DataFrame(results_model1)
    df_results_model2 = pd.DataFrame(results_model2)
    
    df_results_model1.to_csv(results_dir / "summary_model1.csv", index=False)
    df_results_model2.to_csv(results_dir / "summary_model2.csv", index=False)
    
    # Сохраняем в JSON для удобства
    with open(results_dir / "summary_model1.json", 'w') as f:
        json.dump(results_model1, f, indent=2, default=str)
    
    with open(results_dir / "summary_model2.json", 'w') as f:
        json.dump(results_model2, f, indent=2, default=str)
    
    print("\n" + "="*80)
    print("ТЕСТИРОВАНИЕ ЗАВЕРШЕНО")
    print("="*80)
    print(f"Результаты сохранены в: {results_dir.absolute()}")
    print(f"  - Модель 1: {len(results_model1)} датасетов")
    print(f"  - Модель 2: {len(results_model2)} датасетов")
    
    return df_results_model1, df_results_model2

# ============================================================
# ФУНКЦИЯ ДЛЯ ВИЗУАЛИЗАЦИИ СВОДНЫХ РЕЗУЛЬТАТОВ
# ============================================================

def plot_summary_results(df_results_model1, df_results_model2):
    """Создает сводные графики по всем результатам"""
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('Сводные результаты тестирования моделей', fontsize=16, fontweight='bold')
    
    # 1. Сравнение RMSE для I(t)
    ax = axes[0, 0]
    datasets = df_results_model1['dataset'].values
    x = np.arange(len(datasets))
    width = 0.35
    
    ax.bar(x - width/2, df_results_model1['I_RMSE'], width, label='Модель 1', color='red', alpha=0.7)
    ax.bar(x + width/2, df_results_model2['I_RMSE'], width, label='Модель 2', color='blue', alpha=0.7)
    
    ax.set_xlabel('Датасет')
    ax.set_ylabel('RMSE для I(t)')
    ax.set_title('Сравнение точности предсказания I(t)')
    ax.set_xticks(x)
    ax.set_xticklabels([d[:15] + '...' if len(d) > 15 else d for d in datasets], rotation=45, ha='right')
    ax.legend()
    ax.grid(True, alpha=0.3, axis='y')
    
    # 2. Относительная ошибка параметров (для модели 1)
    ax = axes[0, 1]
    if 'beta_rel_error_percent' in df_results_model1.columns:
        for param, color in [('beta', 'orange'), ('gamma', 'green'), ('mu', 'brown')]:
            col_name = f'{param}_rel_error_percent'
            if col_name in df_results_model1.columns:
                ax.plot(df_results_model1['dataset'], df_results_model1[col_name], 
                       'o-', color=color, label=f'{param}', alpha=0.7)
    
    ax.set_xlabel('Датасет')
    ax.set_ylabel('Относительная ошибка (%)')
    ax.set_title('Модель 1: ошибка оценки параметров')
    ax.set_xticklabels([d[:10] + '...' for d in datasets], rotation=45, ha='right')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 3. R² для параметров (модель 2)
    ax = axes[1, 0]
    if 'beta_R2' in df_results_model2.columns:
        ax.plot(df_results_model2['dataset'], df_results_model2['beta_R2'], 
               'o-', color='orange', label='β(t)', alpha=0.7)
    if 'gamma_R2' in df_results_model2.columns:
        ax.plot(df_results_model2['dataset'], df_results_model2['gamma_R2'], 
               'o-', color='green', label='γ(t)', alpha=0.7)
    
    ax.axhline(y=0.0, color='gray', linestyle='--', alpha=0.5)
    ax.axhline(y=1.0, color='gray', linestyle='--', alpha=0.5)
    ax.set_xlabel('Датасет')
    ax.set_ylabel('R²')
    ax.set_title('Модель 2: качество восстановления параметров')
    ax.set_xticklabels([d[:10] + '...' for d in datasets], rotation=45, ha='right')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 4. Сравнение по шумным датасетам
    ax = axes[1, 1]
    noise_datasets = [d for d in datasets if 'noise' in d]
    if noise_datasets:
        noise_idx = [i for i, d in enumerate(datasets) if 'noise' in d]
        
        for idx in noise_idx:
            ax.scatter(idx, df_results_model1.iloc[idx]['I_RMSE'], 
                      color='red', s=100, marker='o', alpha=0.7)
            ax.scatter(idx, df_results_model2.iloc[idx]['I_RMSE'], 
                      color='blue', s=100, marker='s', alpha=0.7)
        
        ax.set_xlabel('Индекс датасета с шумом')
        ax.set_ylabel('RMSE для I(t)')
        ax.set_title('Устойчивость к шуму (красный=мод1, синий=мод2)')
        ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.savefig(results_dir / "summary_comparison.png", dpi=150, bbox_inches='tight')
    plt.show()

# ============================================================
# ЗАПУСК
# ============================================================

if __name__ == "__main__":
    # Запускаем тестирование
    df_res1, df_res2 = run_testing()
    
    # Строим сводные графики
    plot_summary_results(df_res1, df_res2)
    
    print("\n" + "="*80)
    print("ГОТОВО! Все результаты сохранены.")
    print("="*80)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import os
from pathlib import Path
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import json
from datetime import datetime

# ============================================================
# НАСТРОЙКА
# ============================================================

# Определяем девайс
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Фиксируем seed для воспроизводимости
torch.manual_seed(123)
if device.type == 'cuda':
    torch.cuda.manual_seed(123)
np.random.seed(123)
torch.backends.cudnn.deterministic = True

# Создаем папку для результатов
results_dir = Path("synthetic_datasets_results")
results_dir.mkdir(exist_ok=True)

# Создаем подпапки для каждой модели
model1_dir = results_dir / "model1_constant"
model2_dir = results_dir / "model2_temporal"
model1_dir.mkdir(exist_ok=True)
model2_dir.mkdir(exist_ok=True)

# ============================================================
# СПИСОК ДАТАСЕТОВ ДЛЯ ТЕСТИРОВАНИЯ
# ============================================================

datasets = [
    # Основные датасеты
    {"name": "01_baseline_constant", "desc": "Базовый (константы)", "has_noise": False},
    {"name": "02_lockdown_beta_jump", "desc": "Локдаун (скачок β)", "has_noise": False},
    {"name": "03_seasonal_beta_sin", "desc": "Сезонность (плавная β)", "has_noise": False},
    {"name": "04_decaying_beta_trend", "desc": "Затухание (β 0.15→0.1)", "has_noise": False},
    {"name": "05_complex_beta_gamma_dynamics", "desc": "Сложный (β+γ)", "has_noise": False},
    
    # Шумные версии датасета 2
    {"name": "02_lockdown_beta_jump_noise_3pct_seed42", "desc": "Локдаун + шум 3%", "has_noise": True},
    {"name": "02_lockdown_beta_jump_noise_5pct_seed123", "desc": "Локдаун + шум 5%", "has_noise": True},
    {"name": "02_lockdown_beta_jump_noise_10pct_seed456", "desc": "Локдаун + шум 10%", "has_noise": True},
    
    # Шумные версии датасета 5
    {"name": "05_complex_beta_gamma_dynamics_noise_3pct_seed42", "desc": "Сложный + шум 3%", "has_noise": True},
    {"name": "05_complex_beta_gamma_dynamics_noise_5pct_seed123", "desc": "Сложный + шум 5%", "has_noise": True},
    {"name": "05_complex_beta_gamma_dynamics_noise_10pct_seed456", "desc": "Сложный + шум 10%", "has_noise": True},
]

# Истинные параметры для каждого датасета (для расчета ошибок)
true_params = {
    "01_baseline_constant": {"beta": 0.1, "gamma": 0.06, "mu": 0.003, "type": "constant"},
    "02_lockdown_beta_jump": {"beta": "time_varying", "gamma": 0.06, "mu": 0.003, "type": "beta_time"},
    "03_seasonal_beta_sin": {"beta": "time_varying", "gamma": 0.06, "mu": 0.003, "type": "beta_time"},
    "04_decaying_beta_trend": {"beta": "time_varying", "gamma": 0.06, "mu": 0.003, "type": "beta_time"},
    "05_complex_beta_gamma_dynamics": {"beta": "time_varying", "gamma": "time_varying", "mu": 0.003, "type": "both_time"},
}

# Для шумных версий параметры те же, что у исходных
for noise_level in [3, 5, 10]:
    for seed in [42, 123, 456]:
        true_params[f"02_lockdown_beta_jump_noise_{noise_level}pct_seed{seed}"] = true_params["02_lockdown_beta_jump"]
        true_params[f"05_complex_beta_gamma_dynamics_noise_{noise_level}pct_seed{seed}"] = true_params["05_complex_beta_gamma_dynamics"]

# ============================================================
# ФУНКЦИИ ДЛЯ РАСЧЕТА МЕТРИК
# ============================================================

def calculate_metrics(y_true, y_pred):
    """Рассчитывает MSE, RMSE, MAE для временных рядов"""
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    return {"MSE": mse, "RMSE": rmse, "MAE": mae}

def calculate_parameter_error_model1(pred_params, true_params_dict):
    """Относительная ошибка для константных параметров (модель 1)"""
    errors = {}
    for param in ['beta', 'gamma', 'mu']:
        if isinstance(true_params_dict[param], (int, float)):
            true_val = true_params_dict[param]
            pred_val = pred_params[param]
            rel_error = abs(pred_val - true_val) / true_val
            errors[param] = {
                "true": true_val,
                "pred": pred_val,
                "abs_error": abs(pred_val - true_val),
                "rel_error": rel_error,
                "rel_error_percent": rel_error * 100
            }
    return errors

def calculate_parameter_error_model2(df_true, df_pred, param_name):
    """Ошибки для временных параметров (модель 2)"""
    true_vals = df_true[param_name].values
    pred_vals = df_pred[param_name].values
    
    # Обрезаем до одинаковой длины
    min_len = min(len(true_vals), len(pred_vals))
    true_vals = true_vals[:min_len]
    pred_vals = pred_vals[:min_len]
    
    mse = mean_squared_error(true_vals, pred_vals)
    rmse = np.sqrt(mse)
    r2 = r2_score(true_vals, pred_vals)
    
    return {
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2,
        "mean_true": np.mean(true_vals),
        "mean_pred": np.mean(pred_vals)
    }

# ============================================================
# ФУНКЦИИ ДЛЯ ВИЗУАЛИЗАЦИИ
# ============================================================

def plot_model_results(df_true, df_pred_model1, df_pred_model2, dataset_name, save_dir):
    """Создает визуализацию сравнения моделей"""
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    fig.suptitle(f'Сравнение моделей: {dataset_name}', fontsize=16, fontweight='bold')
    
    t = df_true['day'].values
    
    # 1. Инфицированные I(t)
    ax = axes[0, 0]
    ax.plot(t, df_true['I'], 'k-', linewidth=2.5, label='Истинные', alpha=0.8)
    ax.plot(t[:len(df_pred_model1['I'])], df_pred_model1['I'], 'r--', linewidth=2, label='Модель 1 (конст)')
    ax.plot(t[:len(df_pred_model2['I'])], df_pred_model2['I'], 'b--', linewidth=2, label='Модель 2 (врем)')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Инфицированные')
    ax.set_title('Динамика I(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 2. Умершие D(t)
    ax = axes[0, 1]
    ax.plot(t, df_true['D'], 'k-', linewidth=2.5, label='Истинные', alpha=0.8)
    ax.plot(t[:len(df_pred_model1['D'])], df_pred_model1['D'], 'r--', linewidth=2, label='Модель 1')
    ax.plot(t[:len(df_pred_model2['D'])], df_pred_model2['D'], 'b--', linewidth=2, label='Модель 2')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Умершие')
    ax.set_title('Динамика D(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 3. R₀ (если есть в данных)
    ax = axes[0, 2]
    if 'R0' in df_true.columns:
        ax.plot(t, df_true['R0'], 'k-', linewidth=2.5, label='Истинный R₀', alpha=0.8)
    if 'R0' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['R0'])], df_pred_model2['R0'], 'b--', linewidth=2, label='Модель 2 R₀')
    ax.axhline(y=1.0, color='gray', linestyle='--', alpha=0.5)
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('R₀')
    ax.set_title('Репродуктивное число')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 4. Параметр β(t)
    ax = axes[1, 0]
    if 'beta' in df_true.columns:
        ax.plot(t, df_true['beta'], 'k-', linewidth=2.5, label='Истинный β(t)', alpha=0.8)
    if 'beta' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['beta'])], df_pred_model2['beta'], 'b--', linewidth=2, label='Модель 2 β(t)')
    # Для модели 1 - константа
    if 'beta_const' in df_pred_model1.columns:
        beta_const = df_pred_model1['beta_const'].iloc[0]
        ax.axhline(y=beta_const, color='r', linestyle='--', linewidth=2, label=f'Модель 1 β={beta_const:.4f}')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('β')
    ax.set_title('Параметр β(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 5. Параметры γ(t) и μ(t)
    ax = axes[1, 1]
    if 'gamma' in df_true.columns:
        ax.plot(t, df_true['gamma'], 'g-', linewidth=2, label='Истинный γ(t)', alpha=0.6)
    if 'mu' in df_true.columns:
        ax.plot(t, df_true['mu'], 'brown', linewidth=2, label='Истинный μ(t)', alpha=0.6)
    if 'gamma' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['gamma'])], df_pred_model2['gamma'], 'g--', linewidth=2, label='Модель 2 γ(t)')
    if 'mu' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['mu'])], df_pred_model2['mu'], 'orange', linewidth=2, label='Модель 2 μ(t)')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Параметры')
    ax.set_title('Параметры γ(t) и μ(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 6. Ошибки предсказания I(t)
    ax = axes[1, 2]
    min_len = min(len(t), len(df_pred_model1['I']), len(df_pred_model2['I']))
    error1 = np.abs(df_true['I'].values[:min_len] - df_pred_model1['I'][:min_len])
    error2 = np.abs(df_true['I'].values[:min_len] - df_pred_model2['I'][:min_len])
    
    ax.plot(t[:min_len], error1, 'r-', linewidth=1.5, label='Ошибка Модель 1')
    ax.plot(t[:min_len], error2, 'b-', linewidth=1.5, label='Ошибка Модель 2')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('|Ошибка| I(t)')
    ax.set_title('Абсолютная ошибка для I(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.savefig(save_dir / f"{dataset_name}_comparison.png", dpi=150, bbox_inches='tight')
    plt.close()

# ============================================================
# ОСНОВНОЙ ЦИКЛ ТЕСТИРОВАНИЯ
# ============================================================

def run_testing():
    """Запускает тестирование на всех датасетах"""
    
    # Словари для накопления результатов
    results_model1 = []
    results_model2 = []
    
    print("="*80)
    print("НАЧАЛО ТЕСТИРОВАНИЯ МОДЕЛЕЙ")
    print("="*80)
    
    for ds in datasets:
        dataset_name = ds["name"]
        dataset_desc = ds["desc"]
        
        print(f"\n{'='*60}")
        print(f"ТЕСТИРОВАНИЕ: {dataset_name}")
        print(f"Описание: {dataset_desc}")
        print('='*60)
        
        try:
            # ==================================================
            # ЗАГРУЗКА ДАННЫХ
            # ==================================================
            file_path = f"./synthetic_datasets/{dataset_name}.csv"
            if not os.path.exists(file_path):
                print(f"❌ Файл не найден: {file_path}")
                continue
                
            df_true = pd.read_csv(file_path)
            
            # Получаем данные
            S = df_true['S'].values
            I = df_true['I'].values
            D = df_true['D'].values
            R = df_true['R'].values
            timesteps = np.arange(len(S)).astype(float)
            
            # Параметры
            # train_size = min(101, len(timesteps) // 2)  # Обучаем на половине данных
            train_size = 130
            population = 1000
            
            print(f"Всего точек: {len(timesteps)}, Обучающих: {train_size}, Тестовых: {len(timesteps)-train_size}")
            
            # ==================================================
            # МОДЕЛЬ 1: EINN_PINN (константные параметры)
            # ==================================================
            print("\n--- МОДЕЛЬ 1 (константные параметры) ---")
            
            from NEW_PINN.PINN_const import EINN_PINN  # предполагаемый импорт
            
            model1 = EINN_PINN(
                t=timesteps,
                S_data=S,
                I_data=I,
                R_data=R,
                D_data=D,
                population=population,
                train_size=train_size,
                device=device
            )
            
            # Обучаем
            model1.train_model(n_epoch=20000, lambda_data=0.01, lambda_ode=1.0)
            
            # Получаем параметры
            final_params = model1.params.get_params_dict()
            
            # Предсказание
            S_pred1, I_pred1, R_pred1, D_pred1 = model1.predict(timesteps)
            
            # Конвертируем в numpy
            S_pred1 = S_pred1.numpy()
            I_pred1 = I_pred1.numpy()
            R_pred1 = R_pred1.numpy()
            D_pred1 = D_pred1.numpy()
            
            # Создаем DataFrame с предсказаниями модели 1
            df_pred1 = pd.DataFrame({
                'day': timesteps[:len(S_pred1)],
                'S': S_pred1,
                'I': I_pred1,
                'R': R_pred1,
                'D': D_pred1,
                'beta_const': final_params['beta'],
                'gamma_const': final_params['gamma'],
                'mu_const': final_params['mu']
            })
            df_pred1['R0_const'] = df_pred1['beta_const'] / (df_pred1['gamma_const'] + df_pred1['mu_const'])
            
            # ==================================================
            # МОДЕЛЬ 2: Nature_PINN (временные параметры)
            # ==================================================
            print("\n--- МОДЕЛЬ 2 (временные параметры) ---")
            
            from NEW_PINN.PINN_time import Nature_PINN  # предполагаемый импорт
            
            model2 = Nature_PINN(
                t=timesteps,
                S_data=S,
                I_data=I,
                D_data=D,
                R_data=R,
                device=device,
                train_size=train_size,
                population_n=population
            )
            
            # Обучаем
            model2.train(n_epoch=20000)
            
            # Получаем предсказания
            predictions = model2.predict()
            
            # Создаем DataFrame с предсказаниями модели 2
            df_pred2 = pd.DataFrame({
                'day': predictions['t'][:len(predictions['S'])],
                'S': predictions['S'],
                'I': predictions['I'],
                'R': predictions['R'],
                'D': predictions['D'],
                'beta': predictions['beta'],
                'gamma': predictions['gamma'],
                'mu': predictions['mu']
            })
            df_pred2['R0'] = df_pred2['beta'] / (df_pred2['gamma'] + df_pred2['mu'])
            
            # ==================================================
            # РАСЧЕТ МЕТРИК
            # ==================================================
            
            # Обрезаем до одинаковой длины
            min_len = min(len(df_true), len(df_pred1), len(df_pred2))
            df_true_trim = df_true.iloc[:min_len]
            df_pred1_trim = df_pred1.iloc[:min_len]
            df_pred2_trim = df_pred2.iloc[:min_len]
            
            # Метрики для SIRD
            metrics_model1 = {}
            metrics_model2 = {}
            
            for compartment in ['S', 'I', 'R', 'D']:
                true_vals = df_true_trim[compartment].values
                pred1_vals = df_pred1_trim[compartment].values
                pred2_vals = df_pred2_trim[compartment].values
                
                metrics_model1[compartment] = calculate_metrics(true_vals, pred1_vals)
                metrics_model2[compartment] = calculate_metrics(true_vals, pred2_vals)
            
            # Ошибки параметров для модели 1
            param_errors_model1 = {}
            if dataset_name in true_params:
                if true_params[dataset_name]["type"] == "constant":
                    param_errors_model1 = calculate_parameter_error_model1(final_params, true_params[dataset_name])
            
            # Ошибки параметров для модели 2
            param_errors_model2 = {}
            if dataset_name in true_params:
                if 'beta' in df_true.columns and true_params[dataset_name]["beta"] == "time_varying":
                    param_errors_model2['beta'] = calculate_parameter_error_model2(df_true, df_pred2, 'beta')
                if 'gamma' in df_true.columns and true_params[dataset_name]["gamma"] == "time_varying":
                    param_errors_model2['gamma'] = calculate_parameter_error_model2(df_true, df_pred2, 'gamma')
                if 'mu' in df_true.columns and isinstance(true_params[dataset_name]["mu"], (int, float)):
                    param_errors_model2['mu'] = calculate_parameter_error_model2(df_true, df_pred2, 'mu')
            
            # ==================================================
            # СОХРАНЕНИЕ РЕЗУЛЬТАТОВ
            # ==================================================
            
            # Сохраняем предсказания
            df_pred1.to_csv(model1_dir / f"{dataset_name}_predictions.csv", index=False)
            df_pred2.to_csv(model2_dir / f"{dataset_name}_predictions.csv", index=False)
            
            # Сохраняем метрики
            result_entry_model1 = {
                'dataset': dataset_name,
                'description': dataset_desc,
                **{f"{comp}_MSE": metrics_model1[comp]['MSE'] for comp in ['S', 'I', 'R', 'D']},
                **{f"{comp}_RMSE": metrics_model1[comp]['RMSE'] for comp in ['S', 'I', 'R', 'D']},
                **{f"{comp}_MAE": metrics_model1[comp]['MAE'] for comp in ['S', 'I', 'R', 'D']},
                'beta_pred': final_params['beta'],
                'gamma_pred': final_params['gamma'],
                'mu_pred': final_params['mu'],
            }
            
            # Добавляем ошибки параметров если есть
            if param_errors_model1:
                for param, errors in param_errors_model1.items():
                    result_entry_model1[f"{param}_true"] = errors['true']
                    result_entry_model1[f"{param}_rel_error"] = errors['rel_error']
                    result_entry_model1[f"{param}_rel_error_percent"] = errors['rel_error_percent']
            
            results_model1.append(result_entry_model1)
            
            result_entry_model2 = {
                'dataset': dataset_name,
                'description': dataset_desc,
                **{f"{comp}_MSE": metrics_model2[comp]['MSE'] for comp in ['S', 'I', 'R', 'D']},
                **{f"{comp}_RMSE": metrics_model2[comp]['RMSE'] for comp in ['S', 'I', 'R', 'D']},
                **{f"{comp}_MAE": metrics_model2[comp]['MAE'] for comp in ['S', 'I', 'R', 'D']},
            }
            
            # Добавляем ошибки параметров
            for param, errors in param_errors_model2.items():
                result_entry_model2[f"{param}_RMSE"] = errors['RMSE']
                result_entry_model2[f"{param}_R2"] = errors['R2']
            
            results_model2.append(result_entry_model2)
            
            # ==================================================
            # ВИЗУАЛИЗАЦИЯ
            # ==================================================
            
            plot_model_results(df_true, df_pred1, df_pred2, dataset_name, results_dir)
            
            print(f"\n✅ Завершено: {dataset_name}")
            print(f"   Модель 1 - I_RMSE: {metrics_model1['I']['RMSE']:.4f}")
            print(f"   Модель 2 - I_RMSE: {metrics_model2['I']['RMSE']:.4f}")
            
        except Exception as e:
            print(f"❌ Ошибка при обработке {dataset_name}: {str(e)}")
            import traceback
            traceback.print_exc()
            continue
    
    # ==================================================
    # СОХРАНЕНИЕ СВОДНЫХ РЕЗУЛЬТАТОВ
    # ==================================================
    
    # Сохраняем результаты в CSV
    df_results_model1 = pd.DataFrame(results_model1)
    df_results_model2 = pd.DataFrame(results_model2)
    
    df_results_model1.to_csv(results_dir / "summary_model1.csv", index=False)
    df_results_model2.to_csv(results_dir / "summary_model2.csv", index=False)
    
    # Сохраняем в JSON для удобства
    with open(results_dir / "summary_model1.json", 'w') as f:
        json.dump(results_model1, f, indent=2, default=str)
    
    with open(results_dir / "summary_model2.json", 'w') as f:
        json.dump(results_model2, f, indent=2, default=str)
    
    print("\n" + "="*80)
    print("ТЕСТИРОВАНИЕ ЗАВЕРШЕНО")
    print("="*80)
    print(f"Результаты сохранены в: {results_dir.absolute()}")
    print(f"  - Модель 1: {len(results_model1)} датасетов")
    print(f"  - Модель 2: {len(results_model2)} датасетов")
    
    return df_results_model1, df_results_model2

# ============================================================
# ФУНКЦИЯ ДЛЯ ВИЗУАЛИЗАЦИИ СВОДНЫХ РЕЗУЛЬТАТОВ
# ============================================================

def plot_summary_results(df_results_model1, df_results_model2):
    """Создает сводные графики по всем результатам"""
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('Сводные результаты тестирования моделей', fontsize=16, fontweight='bold')
    
    # 1. Сравнение RMSE для I(t)
    ax = axes[0, 0]
    datasets = df_results_model1['dataset'].values
    x = np.arange(len(datasets))
    width = 0.35
    
    ax.bar(x - width/2, df_results_model1['I_RMSE'], width, label='Модель 1', color='red', alpha=0.7)
    ax.bar(x + width/2, df_results_model2['I_RMSE'], width, label='Модель 2', color='blue', alpha=0.7)
    
    ax.set_xlabel('Датасет')
    ax.set_ylabel('RMSE для I(t)')
    ax.set_title('Сравнение точности предсказания I(t)')
    ax.set_xticks(x)
    ax.set_xticklabels([d[:15] + '...' if len(d) > 15 else d for d in datasets], rotation=45, ha='right')
    ax.legend()
    ax.grid(True, alpha=0.3, axis='y')
    
    # 2. Относительная ошибка параметров (для модели 1)
    ax = axes[0, 1]
    if 'beta_rel_error_percent' in df_results_model1.columns:
        for param, color in [('beta', 'orange'), ('gamma', 'green'), ('mu', 'brown')]:
            col_name = f'{param}_rel_error_percent'
            if col_name in df_results_model1.columns:
                ax.plot(df_results_model1['dataset'], df_results_model1[col_name], 
                       'o-', color=color, label=f'{param}', alpha=0.7)
    
    ax.set_xlabel('Датасет')
    ax.set_ylabel('Относительная ошибка (%)')
    ax.set_title('Модель 1: ошибка оценки параметров')
    ax.set_xticklabels([d[:10] + '...' for d in datasets], rotation=45, ha='right')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 3. R² для параметров (модель 2)
    ax = axes[1, 0]
    if 'beta_R2' in df_results_model2.columns:
        ax.plot(df_results_model2['dataset'], df_results_model2['beta_R2'], 
               'o-', color='orange', label='β(t)', alpha=0.7)
    if 'gamma_R2' in df_results_model2.columns:
        ax.plot(df_results_model2['dataset'], df_results_model2['gamma_R2'], 
               'o-', color='green', label='γ(t)', alpha=0.7)
    
    ax.axhline(y=0.0, color='gray', linestyle='--', alpha=0.5)
    ax.axhline(y=1.0, color='gray', linestyle='--', alpha=0.5)
    ax.set_xlabel('Датасет')
    ax.set_ylabel('R²')
    ax.set_title('Модель 2: качество восстановления параметров')
    ax.set_xticklabels([d[:10] + '...' for d in datasets], rotation=45, ha='right')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 4. Сравнение по шумным датасетам
    ax = axes[1, 1]
    noise_datasets = [d for d in datasets if 'noise' in d]
    if noise_datasets:
        noise_idx = [i for i, d in enumerate(datasets) if 'noise' in d]
        
        for idx in noise_idx:
            ax.scatter(idx, df_results_model1.iloc[idx]['I_RMSE'], 
                      color='red', s=100, marker='o', alpha=0.7)
            ax.scatter(idx, df_results_model2.iloc[idx]['I_RMSE'], 
                      color='blue', s=100, marker='s', alpha=0.7)
        
        ax.set_xlabel('Индекс датасета с шумом')
        ax.set_ylabel('RMSE для I(t)')
        ax.set_title('Устойчивость к шуму (красный=мод1, синий=мод2)')
        ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.savefig(results_dir / "summary_comparison.png", dpi=150, bbox_inches='tight')
    plt.show()

# ============================================================
# ЗАПУСК
# ============================================================

if __name__ == "__main__":
    # Запускаем тестирование
    df_res1, df_res2 = run_testing()
    
    # Строим сводные графики
    plot_summary_results(df_res1, df_res2)
    
    print("\n" + "="*80)
    print("ГОТОВО! Все результаты сохранены.")
    print("="*80)
```
