```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import os
from pathlib import Path
from sklearn.metrics import mean_squared_error, mean_absolute_error
import json

# ============================================================
# НАСТРОЙКА
# ============================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Фиксируем seed
torch.manual_seed(123)
if device.type == 'cuda':
    torch.cuda.manual_seed(123)
np.random.seed(123)
torch.backends.cudnn.deterministic = True

# Создаем папку для результатов
results_dir = Path("real_datasets_results")
results_dir.mkdir(exist_ok=True)

# ============================================================
# ЗАГРУЗКА РЕАЛЬНЫХ ДАННЫХ
# ============================================================

print("="*60)
print("ЗАГРУЗКА РЕАЛЬНОГО ДАТАСЕТА")
print("="*60)

# Загружаем данные
df_real = pd.read_csv('real_datasets\PINN-COVID-Italy.csv')

# Переименовываем колонки для удобства (если нужно)
# df_real.columns = ['I', 'D', 'S', 'R'] # RUS
df_real.columns = ['S', 'I', 'R', 'D'] # ITALY

# Получаем данные
timesteps = np.arange(len(S)).astype(float)
df_real['t'] = timesteps
S = df_real['S'].values
I = df_real['I'].values
D = df_real['D'].values
R = df_real['R'].values

population = S[0] + I[0] + R[0] + D[0]  # Общая популяция в начале

print(f"Всего точек данных: {len(timesteps)}")
print(f"Население: {population}")
print(f"Диапазон времени: {timesteps[0]} - {timesteps[-1]} дней")

# ============================================================
# РАЗМЕРЫ ОБУЧАЮЩЕЙ ВЫБОРКИ ДЛЯ ТЕСТИРОВАНИЯ
# ============================================================

# train_sizes = [120, 150, 180, 200, 220] # rus

train_sizes = [70, 100, 250, 280, 300] # Italy

# ============================================================
# ФУНКЦИИ ДЛЯ РАСЧЕТА МЕТРИК
# ============================================================

def calculate_metrics(y_true, y_pred):
    """Рассчитывает MSE, RMSE, MAE для временных рядов"""
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    return {"MSE": mse, "RMSE": rmse, "MAE": mae}

# ============================================================
# ФУНКЦИЯ ДЛЯ ВИЗУАЛИЗАЦИИ С ПУНКТИРНОЙ ЛИНИЕЙ
# ============================================================

def plot_real_data_results(df_true, df_pred_model1, df_pred_model2, train_size, save_dir):
    """Создает визуализацию для реальных данных с пунктирной линией train/test"""
    
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    fig.suptitle(f'Реальные данные: COVID-19 (train_size={train_size})', 
                 fontsize=16, fontweight='bold')
    
    t = df_true['t'].values
    
    # Вертикальная линия, разделяющая train и test
    train_cutoff = t[train_size - 1] if train_size < len(t) else t[-1]
    
    # 1. Инфицированные I(t)
    ax = axes[0, 0]
    ax.plot(t, df_true['I'], 'k-', linewidth=2.5, label='Реальные данные', alpha=0.8)
    ax.plot(t[:len(df_pred_model1['I'])], df_pred_model1['I'], 'r--', linewidth=2, label='Модель 1 (конст)')
    ax.plot(t[:len(df_pred_model2['I'])], df_pred_model2['I'], 'b--', linewidth=2, label='Модель 2 (врем)')
    ax.axvline(x=train_cutoff, color='gray', linestyle=':', linewidth=2, 
               label=f'Граница train/test (день {train_cutoff:.0f})')
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Инфицированные')
    ax.set_title('Динамика I(t)')
    ax.legend(loc='upper left', fontsize=8)
    ax.grid(True, alpha=0.3)
    
    # 2. Умершие D(t)
    ax = axes[0, 1]
    ax.plot(t, df_true['D'], 'k-', linewidth=2.5, label='Реальные данные', alpha=0.8)
    ax.plot(t[:len(df_pred_model1['D'])], df_pred_model1['D'], 'r--', linewidth=2, label='Модель 1')
    ax.plot(t[:len(df_pred_model2['D'])], df_pred_model2['D'], 'b--', linewidth=2, label='Модель 2')
    ax.axvline(x=train_cutoff, color='gray', linestyle=':', linewidth=2)
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Умершие')
    ax.set_title('Динамика D(t)')
    ax.legend(loc='upper left', fontsize=8)
    ax.grid(True, alpha=0.3)
    
    # 3. Выздоровевшие R(t)
    ax = axes[0, 2]
    ax.plot(t, df_true['R'], 'k-', linewidth=2.5, label='Реальные данные', alpha=0.8)
    ax.plot(t[:len(df_pred_model1['R'])], df_pred_model1['R'], 'r--', linewidth=2, label='Модель 1')
    ax.plot(t[:len(df_pred_model2['R'])], df_pred_model2['R'], 'b--', linewidth=2, label='Модель 2')
    ax.axvline(x=train_cutoff, color='gray', linestyle=':', linewidth=2)
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Выздоровевшие')
    ax.set_title('Динамика R(t)')
    ax.legend(loc='upper left', fontsize=8)
    ax.grid(True, alpha=0.3)
    
    # 4. Восприимчивые S(t)
    ax = axes[1, 0]
    ax.plot(t, df_true['S'], 'k-', linewidth=2.5, label='Реальные данные', alpha=0.8)
    ax.plot(t[:len(df_pred_model1['S'])], df_pred_model1['S'], 'r--', linewidth=2, label='Модель 1')
    ax.plot(t[:len(df_pred_model2['S'])], df_pred_model2['S'], 'b--', linewidth=2, label='Модель 2')
    ax.axvline(x=train_cutoff, color='gray', linestyle=':', linewidth=2)
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Восприимчивые')
    ax.set_title('Динамика S(t)')
    ax.legend(loc='upper right', fontsize=8)
    ax.grid(True, alpha=0.3)
    
    # 5. Параметры модели 2 (β, γ, μ)
    ax = axes[1, 1]
    if 'beta' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['beta'])], df_pred_model2['beta'], 
                'orange', linewidth=2, label='β(t)')
    if 'gamma' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['gamma'])], df_pred_model2['gamma'], 
                'green', linewidth=2, label='γ(t)')
    if 'mu' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['mu'])], df_pred_model2['mu'], 
                'brown', linewidth=2, label='μ(t)')
    ax.axvline(x=train_cutoff, color='gray', linestyle=':', linewidth=2)
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('Значения параметров')
    ax.set_title('Параметры модели 2 (временные)')
    ax.legend(loc='upper right', fontsize=8)
    ax.grid(True, alpha=0.3)
    
    # 6. Репродуктивное число R₀ (для модели 2)
    ax = axes[1, 2]
    if 'R0' in df_pred_model2.columns:
        ax.plot(t[:len(df_pred_model2['R0'])], df_pred_model2['R0'], 
                'purple', linewidth=2, label='R₀(t) - модель 2')
    # Добавляем константное R₀ для модели 1
    if 'R0_const' in df_pred_model1.columns:
        r0_const = df_pred_model1['R0_const'].iloc[0]
        ax.axhline(y=r0_const, color='red', linestyle='--', linewidth=2, 
                   label=f'R₀ const = {r0_const:.3f} (мод 1)')
    ax.axhline(y=1.0, color='black', linestyle='-', linewidth=1, alpha=0.5)
    ax.axvline(x=train_cutoff, color='gray', linestyle=':', linewidth=2)
    ax.set_xlabel('Время (дни)')
    ax.set_ylabel('R₀')
    ax.set_title('Репродуктивное число')
    ax.legend(loc='upper right', fontsize=8)
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.savefig(save_dir / f"real_data_train{train_size}_comparison.png", 
                dpi=150, bbox_inches='tight')
    plt.close()

# ============================================================
# ОСНОВНОЙ ЦИКЛ ТЕСТИРОВАНИЯ
# ============================================================

def run_real_data_testing():
    """Запускает тестирование на реальных данных"""
    
    results = []
    
    print("="*80)
    print("НАЧАЛО ТЕСТИРОВАНИЯ НА РЕАЛЬНЫХ ДАННЫХ")
    print("="*80)
    
    for train_size in train_sizes:
        print(f"\n{'='*60}")
        print(f"ТЕСТИРОВАНИЕ: train_size = {train_size}")
        print('='*60)
        
        try:
            # Проверяем, что train_size не превышает длину данных
            if train_size > len(timesteps):
                print(f"❌ train_size ({train_size}) больше длины данных ({len(timesteps)})")
                continue
            
            # ==================================================
            # МОДЕЛЬ 1: EINN_PINN (константные параметры)
            # ==================================================
            print("\n--- МОДЕЛЬ 1 (константные параметры) ---")
            
            from NEW_PINN.PINN_const import EINN_PINN
            
            model1 = EINN_PINN(
                t=timesteps,
                S_data=S,
                I_data=I,
                R_data=R,
                D_data=D,
                population=population,
                train_size=train_size,
                device=device
            )
            
            # Обучаем
            model1.train_model(n_epoch=20000, lambda_data=1.0, lambda_ode=1.0, lambda_ic=1.0, lambda_bc=1.0)
            
            # Получаем параметры
            final_params = model1.params.get_params_dict()
            
            # Предсказание
            S_pred1, I_pred1, R_pred1, D_pred1 = model1.predict(timesteps)
            
            # Конвертируем в numpy
            S_pred1 = S_pred1.numpy()
            I_pred1 = I_pred1.numpy()
            R_pred1 = R_pred1.numpy()
            D_pred1 = D_pred1.numpy()
            
            # Создаем DataFrame с предсказаниями модели 1
            df_pred1 = pd.DataFrame({
                't': timesteps[:len(S_pred1)],
                'S': S_pred1,
                'I': I_pred1,
                'R': R_pred1,
                'D': D_pred1,
                'beta_const': final_params['beta'],
                'gamma_const': final_params['gamma'],
                'mu_const': final_params['mu']
            })
            df_pred1['R0_const'] = df_pred1['beta_const'] / (df_pred1['gamma_const'] + df_pred1['mu_const'])
            
            # ==================================================
            # МОДЕЛЬ 2: Nature_PINN (временные параметры)
            # ==================================================
            print("\n--- МОДЕЛЬ 2 (временные параметры) ---")
            
            from NEW_PINN.PINN_time import Nature_PINN
            
            model2 = Nature_PINN(
                t=timesteps,
                S_data=S,
                I_data=I,
                D_data=D,
                R_data=R,
                device=device,
                train_size=train_size,
                population_n=population
            )
            
            # Обучаем
            model2.train(n_epoch=20000, lambda_data=1.0, lambda_ode=1000.0)
            
            # Получаем предсказания
            predictions = model2.predict()
            
            # Создаем DataFrame с предсказаниями модели 2
            df_pred2 = pd.DataFrame({
                't': timesteps[:len(predictions['S'])],
                'S': predictions['S'],
                'I': predictions['I'],
                'R': predictions['R'],
                'D': predictions['D'],
                'beta': predictions['beta'],
                'gamma': predictions['gamma'],
                'mu': predictions['mu']
            })
            df_pred2['R0'] = df_pred2['beta'] / (df_pred2['gamma'] + df_pred2['mu'])
            
            # ==================================================
            # РАСЧЕТ МЕТРИК
            # ==================================================
            
            # Обрезаем до одинаковой длины
            min_len = min(len(df_real), len(df_pred1), len(df_pred2))
            df_real_trim = df_real.iloc[:min_len]
            df_pred1_trim = df_pred1.iloc[:min_len]
            df_pred2_trim = df_pred2.iloc[:min_len]
            
            # Метрики для SIRD
            metrics_model1 = {}
            metrics_model2 = {}
            
            # Метрики на всей выборке
            for compartment in ['S', 'I', 'R', 'D']:
                true_vals = df_real_trim[compartment].values
                pred1_vals = df_pred1_trim[compartment].values
                pred2_vals = df_pred2_trim[compartment].values
                
                metrics_model1[compartment] = calculate_metrics(true_vals, pred1_vals)
                metrics_model2[compartment] = calculate_metrics(true_vals, pred2_vals)
            
            # Метрики только на тестовой части (после train_size)
            if train_size < min_len:
                test_metrics_model1 = {}
                test_metrics_model2 = {}
                
                for compartment in ['S', 'I', 'R', 'D']:
                    true_vals = df_real_trim[compartment].values[train_size:]
                    pred1_vals = df_pred1_trim[compartment].values[train_size:]
                    pred2_vals = df_pred2_trim[compartment].values[train_size:]
                    
                    if len(true_vals) > 0:
                        test_metrics_model1[compartment] = calculate_metrics(true_vals, pred1_vals)
                        test_metrics_model2[compartment] = calculate_metrics(true_vals, pred2_vals)
            else:
                test_metrics_model1 = None
                test_metrics_model2 = None
            
            # ==================================================
            # СОХРАНЕНИЕ РЕЗУЛЬТАТОВ
            # ==================================================
            
            # Сохраняем предсказания
            df_pred1.to_csv(results_dir / f"real_data_model1_train{train_size}.csv", index=False)
            df_pred2.to_csv(results_dir / f"real_data_model2_train{train_size}.csv", index=False)
            
            # Сохраняем метрики
            result_entry = {
                'train_size': train_size,
                'test_size': min_len - train_size,
                
                # Модель 1
                'model1_I_RMSE': metrics_model1['I']['RMSE'],
                'model1_I_MAE': metrics_model1['I']['MAE'],
                'model1_D_RMSE': metrics_model1['D']['RMSE'],
                'model1_D_MAE': metrics_model1['D']['MAE'],
                'model1_beta': final_params['beta'],
                'model1_gamma': final_params['gamma'],
                'model1_mu': final_params['mu'],
                'model1_R0': final_params['beta'] / (final_params['gamma'] + final_params['mu']),
                
                # Модель 2
                'model2_I_RMSE': metrics_model2['I']['RMSE'],
                'model2_I_MAE': metrics_model2['I']['MAE'],
                'model2_D_RMSE': metrics_model2['D']['RMSE'],
                'model2_D_MAE': metrics_model2['D']['MAE'],
                'model2_beta_mean': np.mean(df_pred2['beta']),
                'model2_gamma_mean': np.mean(df_pred2['gamma']),
                'model2_mu_mean': np.mean(df_pred2['mu']),
                'model2_R0_mean': np.mean(df_pred2['R0']),
            }
            
            # Добавляем тестовые метрики если есть
            if test_metrics_model1 and test_metrics_model2:
                result_entry.update({
                    'model1_test_I_RMSE': test_metrics_model1['I']['RMSE'],
                    'model1_test_D_RMSE': test_metrics_model1['D']['RMSE'],
                    'model2_test_I_RMSE': test_metrics_model2['I']['RMSE'],
                    'model2_test_D_RMSE': test_metrics_model2['D']['RMSE'],
                })
            
            results.append(result_entry)
            
            # ==================================================
            # ВИЗУАЛИЗАЦИЯ
            # ==================================================
            
            plot_real_data_results(df_real, df_pred1, df_pred2, train_size, results_dir)
            
            print(f"\n✅ Завершено: train_size={train_size}")
            print(f"   Модель 1 - I_RMSE: {metrics_model1['I']['RMSE']:.2f}, R₀={result_entry['model1_R0']:.3f}")
            print(f"   Модель 2 - I_RMSE: {metrics_model2['I']['RMSE']:.2f}, R₀_mean={result_entry['model2_R0_mean']:.3f}")
            
        except Exception as e:
            print(f"❌ Ошибка при train_size={train_size}: {str(e)}")
            import traceback
            traceback.print_exc()
            continue
    
    # ==================================================
    # СОХРАНЕНИЕ СВОДНЫХ РЕЗУЛЬТАТОВ
    # ==================================================
    
    df_results = pd.DataFrame(results)
    df_results.to_csv(results_dir / "real_data_summary.csv", index=False)
    
    with open(results_dir / "real_data_summary.json", 'w') as f:
        json.dump(results, f, indent=2, default=str)
    
    print("\n" + "="*80)
    print("ТЕСТИРОВАНИЕ НА РЕАЛЬНЫХ ДАННЫХ ЗАВЕРШЕНО")
    print("="*80)
    print(f"Результаты сохранены в: {results_dir.absolute()}")
    
    return df_results

# ============================================================
# ФУНКЦИЯ ДЛЯ ВИЗУАЛИЗАЦИИ СВОДНЫХ РЕЗУЛЬТАТОВ
# ============================================================

def plot_real_data_summary(df_results):
    """Создает сводные графики по результатам на реальных данных"""
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle('Реальные данные: сводные результаты по размерам обучающей выборки', 
                 fontsize=16, fontweight='bold')
    
    train_sizes = df_results['train_size'].values
    
    # 1. Сравнение RMSE для I(t)
    ax = axes[0, 0]
    ax.plot(train_sizes, df_results['model1_I_RMSE'], 'ro-', linewidth=2, markersize=8, label='Модель 1')
    ax.plot(train_sizes, df_results['model2_I_RMSE'], 'bo-', linewidth=2, markersize=8, label='Модель 2')
    ax.set_xlabel('Размер обучающей выборки (дней)')
    ax.set_ylabel('RMSE для I(t)')
    ax.set_title('Точность предсказания I(t)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 2. Сравнение RMSE для D(t)
    ax = axes[0, 1]
    ax.plot(train_sizes, df_results['model1_D_RMSE'], 'ro-', linewidth=2, markersize=8, label='Модель 1')
    ax.plot(train_sizes, df_results['model2_D_RMSE'], 'bo-', linewidth=2, markersize=8, label='Модель 2')
    ax.set_xlabel('Размер обучающей выборки (дней)')
    ax.set_ylabel('RMSE для D(t)')
    ax.set_title('Точность предсказания смертности')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 3. Оценка R₀
    ax = axes[1, 0]
    ax.plot(train_sizes, df_results['model1_R0'], 'ro-', linewidth=2, markersize=8, label='Модель 1 (конст)')
    ax.plot(train_sizes, df_results['model2_R0_mean'], 'bo-', linewidth=2, markersize=8, label='Модель 2 (средний)')
    ax.axhline(y=1.0, color='gray', linestyle='--', alpha=0.7)
    ax.set_xlabel('Размер обучающей выборки (дней)')
    ax.set_ylabel('R₀')
    ax.set_title('Оценка репродуктивного числа')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # 4. Оценка параметров модели 1
    ax = axes[1, 1]
    ax.plot(train_sizes, df_results['model1_beta'], 'orange', marker='o', linewidth=2, label='β')
    ax.plot(train_sizes, df_results['model1_gamma'], 'green', marker='s', linewidth=2, label='γ')
    ax.plot(train_sizes, df_results['model1_mu'], 'brown', marker='^', linewidth=2, label='μ')
    ax.set_xlabel('Размер обучающей выборки (дней)')
    ax.set_ylabel('Значения параметров')
    ax.set_title('Модель 1: оценка параметров')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.savefig(results_dir / "real_data_summary.png", dpi=150, bbox_inches='tight')
    plt.show()

# ============================================================
# ЗАПУСК
# ============================================================

if __name__ == "__main__":
    # Запускаем тестирование на реальных данных
    df_results = run_real_data_testing()
    
    # Строим сводные графики
    if len(df_results) > 0:
        plot_real_data_summary(df_results)
    
    print("\n" + "="*80)
    print("ГОТОВО! Все результаты сохранены.")
    print("="*80)
```
